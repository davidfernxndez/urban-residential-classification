# 1. Motivación y aplicación práctica

La selección del modelo que mejor equilibra capacidad predictiva e interpretabilidad constituye únicamente el punto de partida para su aplicación práctica. Para que dicho modelo pueda emplearse como herramienta de apoyo a la toma de decisiones, resulta necesario disponer de un entorno que permita acceder a sus funcionalidades y utilizarlas de forma sencilla en un contexto de aplicación real. Con este propósito, se ha desarrollado una plataforma interactiva que implementa el modelo *XGBoost* y el mecanismo de explicabilidad *SHAP*, trasladando los resultados obtenidos en este trabajo a una herramienta accesible para investigadores y profesionales del ámbito urbanístico y sociológico.

La plataforma implementa dos funcionalidades principales:

* **Análisis del censo de complejos residenciales del área metropolitana de Granada.** Permite consultar de forma dinámica la predicción del modelo y la explicación asociada a cada complejo residencial de la base de datos. Esta funcionalidad facilita a los expertos del dominio la comprensión de los patrones aprendidos por el sistema para caracterizar los distintos grados de cerramiento.

* **Estudio de nuevos desarrollos urbanísticos.** Permite configurar manualmente las características morfológicas de un complejo residencial y obtener de forma inmediata la predicción del grado de cerramiento junto con la explicación asociada. Esta funcionalidad extiende el ámbito de aplicación del sistema al análisis de nuevos complejos residenciales, tanto existentes como hipotéticos, facilitando el estudio del impacto que diferentes configuraciones urbanísticas tienen sobre el grado de cerramiento.

La plataforma constituye así una herramienta de apoyo al análisis sociológico de la fragmentación urbana y materializa el objetivo de transferencia de conocimiento planteado en este trabajo, poniendo a disposición de los expertos un entorno interactivo para consultar, interpretar y aplicar el modelo desarrollado en contextos reales de estudio y planificación urbanística.

<a id="arquitectura"></a>
# 2. Arquitectura y funcionamiento

La plataforma ha sido desarrollada mediante la librería *Streamlit*, que permite integrar en un único entorno la interfaz de usuario, la gestión de datos y la ejecución de modelos de aprendizaje automático.

La arquitectura implementada define dos flujos de ejecución diferenciados, correspondientes a las funcionalidades principales de la aplicación: el análisis de complejos residenciales existentes y la simulación de nuevos desarrollos urbanísticos. La siguiente figura muestra el diagrama de bloques que representa el flujo de funcionamiento seguido por la aplicación en cada uno de estos escenarios.

<img src="../images/app_demo/arquitectura.png" alt="Arquitectura funcional de la plataforma." width="60%">

Al iniciarse, la aplicación carga en memoria los recursos necesarios para su funcionamiento:

* **Predicciones y explicaciones precomputadas.** Se almacenan las predicciones y los valores *SHAP* precomputados para todos los complejos residenciales de la base de datos, obtenidos mediante el enfoque *out-of-fold* descrito en la Sección. De este modo, cada predicción y su correspondiente explicación proceden de un modelo que no ha sido entrenado con la observación analizada.

* **Modelo de inferencia.** Se carga el modelo *XGBoost* entrenado sobre la totalidad de los datos disponibles, tal y como se describe en la Sección. Este modelo constituye el motor de inferencia de la plataforma y se utiliza para generar, en tiempo real, la predicción del grado de cerramiento y su explicación local para nuevos complejos residenciales definidos por el usuario.

## 3.1 Explorador de complejos residenciales

Esta interfaz implementa la funcionalidad de exploración del censo de complejos residenciales del área metropolitana de Granada. Para ello, se ha desarrollado un mapa interactivo mediante la librería *Folium*, en el que se representan geográficamente todos los complejos residenciales disponibles en la base de datos, permitiendo al usuario realizar búsquedas de forma dinámica y fluida.

La siguiente figura muestra la vista inicial de la interfaz, donde se introduce al usuario en el funcionamiento del explorador y se presenta el estado inicial del mapa. Con el objetivo de facilitar la navegación, los complejos residenciales se agrupan automáticamente en función de su proximidad geográfica, adaptando dinámicamente el nivel de detalle al zoom seleccionado por el usuario.

<img src="../images/app_demo/mapa_principal.jpg" alt="mapa_principal" width="80%">

A continuación se presenta una zona ampliada del mapa, donde cada complejo residencial se representa mediante un marcador cuyo color corresponde al grado de cerramiento predicho por el modelo.

<img src="../images/app_demo/mapa_detalle.png" alt="mapa_detalle" width="80%">

Al situar el cursor sobre un marcador, se despliega un cuadro informativo con el identificador del complejo residencial (CC), el municipio al que pertenece y el grado de cerramiento predicho, como se muestra en la siguiente figura.

<img src="../images/app_demo/mapa_click.png" alt="mapa_click" width="80%">

Para implementar en la plataforma el mecanismo de interpretabilidad local presentado en el *notebook* [6_Local_interp_analysis.ipynb](6_Local_interp_analysis.ipynb), se ha desarrollado un panel de explicabilidad situado debajo del mapa, que permanece vacío hasta que el usuario selecciona el complejo residencial que desea analizar haciendo *clic* sobre su marcador. Tras esta acción, el panel se actualiza dinámicamente mostrando la información de predicción y explicabilidad correspondiente al complejo residencial seleccionado, como se muestra en la siguiente figura.

<img src="../images/app_demo/panel_explicabilidad_8707012.png" alt="panel_explicabilidad_8707012" width="100%">

Este panel integra la información asociada a la predicción y las características morfológicas del complejo residencial, situadas en la parte izquierda, junto con el gráfico de cascada (*waterfall plot*) correspondiente al grado de cerramiento predicho. Bajo este gráfico se incorpora una explicación textual basada en las variables con mayor influencia en la decisión del modelo.

Las características morfológicas se organizan mediante menús desplegables que permiten consultar la configuración urbanística del complejo residencial agrupada en los bloques semánticos definidos. La siguiente figura muestra algunos de estos menús desplegados para el complejo residencial analizado en la figura anterior.

<img src="../images/app_demo/caracteristicas_8707012.png" alt="caracteristicas_8707012" width="40%">

Debajo del panel de explicabilidad local se ha incorporado un menú desplegable, inicialmente cerrado, denominado *Análisis complementario de grados de cerramiento alternativos*. Este apartado permite analizar la distribución de probabilidades estimada por el modelo y consultar los gráficos de cascada del resto de grados de cerramiento, como se muestra en la siguiente figura.

<img src="../images/app_demo/detallado_8707012_A.png" alt="detallado_8707012_A" width="90%">
<img src="../images/app_demo/detallado_8707012_B.png" alt="detallado_8707012_B" width="90%">

En conjunto, esta interfaz proporciona un entorno interactivo que permite explorar de forma sencilla y dinámica la distribución cartográfica de los complejos residenciales y consultar tanto las predicciones del modelo como las explicaciones asociadas.

Las herramientas de interpretabilidad incorporadas permiten comprender tanto los factores que han contribuido a la predicción del grado de cerramiento predicho, como los que han reducido la salida del modelo hacia el resto de categorías. Esta visión completa del proceso de decisión se complementa con la posibilidad de consultar la configuración urbanística del complejo residencial de forma estructurada, lo que facilita la interpretación de la relación entre sus características morfológicas y la clasificación obtenida.

## 3.2 Simulador de nuevos escenarios

Esta interfaz implementa la funcionalidad de inferencia sobre nuevos complejos residenciales definidos por el usuario.

Con el objetivo de facilitar la configuración de la morfología urbana del complejo residencial, se ha desarrollado un formulario organizado en bloques semánticos. Cada característica puede configurarse mediante la selección o deselección de un botón, indicando la presencia o ausencia del correspondiente elemento morfológico. La siguiente figura muestra la vista principal de esta interfaz junto con un ejemplo de configuración de algunos de estos bloques. En aquellos casos en los que las variables son mutuamente excluyentes, como los bloques *Distancia al núcleo urbano principal* y *Uso de la vía pública*, la selección queda restringida a una única opción.

<img src="../images/app_demo/formulario_estructurales.png" alt="formulario_estructurales" width="90%">
<img src="../images/app_demo/formulario_distancia.png" alt="formulario_distancia" width="90%">
<img src="../images/app_demo/formulario_cerramiento.png" alt="formulario_cerramiento" width="90%">

Una vez configurado el complejo residencial que se desea evaluar, el usuario puede pulsar el botón Predecir para ejecutar la inferencia con el modelo XGBoost integrado en la plataforma y obtener la predicción junto con su análisis de explicabilidad local.

Como resultado, se genera un panel como el mostrado en la siguiente figura, que incluye el grado de cerramiento predicho, la distribución de probabilidades, el gráfico de cascada asociado y una explicación textual basada en las variables con mayor influencia en la decisión del modelo.

<img src="../images/app_demo/prediccion.png" alt="prediccion" width="90%">

Al igual que en la interfaz anterior, el panel incorpora un menú desplegable adicional situado en la zona inferior, que permite consultar los gráficos de cascada asociados a los grados de cerramiento alternativos, como se muestra en la siguiente figura.

<img src="../images/app_demo/prediccion_alternativa_A.png" alt="prediccion_alternativa_A" width="90%">
<img src="../images/app_demo/prediccion_alternativa_B.png" alt="prediccion_alternativa_B" width="90%">

Esta interfaz proporciona un entorno interactivo para la simulación y el análisis de nuevos desarrollos urbanísticos, permitiendo evaluar cómo diferentes configuraciones de la morfología urbana condicionan el grado de cerramiento inferido por el modelo. 

El usuario puede modificar de forma iterativa la presencia o ausencia de elementos urbanísticos y observar el efecto de estos cambios tanto sobre la predicción y la distribución de probabilidades como sobre la influencia de cada variable en la decisión del modelo. De este modo, la plataforma trasciende el uso meramente predictivo y se convierte en una herramienta de apoyo para la exploración de escenarios hipotéticos, facilitando el análisis del impacto potencial de distintas decisiones de diseño urbano sobre el fenómeno de la fragmentación urbana.
